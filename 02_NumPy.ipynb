```python
sales = np.array([
    [100, 200, 300],
    [150, 250, 350],
    [200, 300, 400]
])

print(sales)
```

```python
print(sales[2,2])
```

```python
print(sales[2,1])
print(sales[1,2])
print(sales[1])
print(sales[:, 1])
print(sales[:1, 1])
```

```python
print(np.sum(sales, axis=0))
```

```python
print(np.sum(sales, axis=1))
```

```python
print(np.mean(sales, axis=0))
print(np.mean(sales, axis=1))
```

```python
print(np.mean(sales[2]))
print(np.mean(sales[:,2])) #mean of only third column
print(np.max(sales, axis=0))
```

```python
np.mean(sales[:, 0])  # mean of 1st column
np.mean(sales[:, 1])  # mean of 2nd column
np.mean(sales[:, 2])  # mean of 3rd column
```
