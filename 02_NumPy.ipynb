```python
data = np.array([100, 200, np.nan, 400, 500])

print(data)
print(np.mean(data))
print(np.nanmean(data))
```

```python
print(data)
print(type(data))
print(np.mean(data))
print(np.nanmean(data))
```

```python
print(np.isnan(data))
```

```python
print(data[np.isnan(data)])
```

```python
valid_data = data[~np.isnan(data)]

print(valid_data)

print("Valid values:", valid_data)
print("Average:", np.mean(valid_data))
print("Total:", np.sum(valid_data))
```

```python
data_filled = np.nan_to_num(data, nan=0)

print(data_filled)
#Important professional rule. Don't automatically replace missing data with 0 or the average.

#For example, if a bank transaction amount is missing, replacing it with 0 could create a false transaction. You first need to understand why the data is missing.

#A good workflow is: Detect → Investigate → Decide → Replace/remove → Verify

#This is exactly the kind of data-quality problem you'll encounter with real bank, telecom, government, and business datasets.
```

```python
data = np.array([100, 200, np.nan, 400, 500])
data_filled = np.nan_to_num(data, nan=300)

print(data_filled)
```
